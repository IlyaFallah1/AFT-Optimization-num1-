```python
import warnings
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

warnings.filterwarnings("ignore")

# =====================================================================
# بخش اول: سورس کد گیت‌هاب هر ۵ تابع موجود در برگه (استاندارد CEC 2019)
# =====================================================================


# ۱. سورس تابع Storn's Chebyshev Polynomial Fitting (D=9, بازه -8192 تا 8192)
def cec19_func1_chebyshev(x):
    D = len(x)
    m = 32
    d = 72.661
    i = np.arange(1, D + 1)

    u = np.sum(x * (1.2 ** (D - i)))
    P1 = (u - d) ** 2 if u < d else 0.0

    v = np.sum(x * ((-1.2) ** (D - i)))
    P2 = (v - d) ** 2 if v < d else 0.0

    k = np.arange(m + 1)[:, None]
    wk = np.sum(x * ((2.0 * k / m - 1.0) ** (D - i)), axis=1)
    pk = np.where(
        wk > 1.0, (wk - 1.0) ** 2, np.where(wk < -1.0, (wk + 1.0) ** 2, 0.0)
    )
    P3 = np.sum(pk)

    return 1.0 + P1 + P2 + P3


# ۲. سورس تابع Inverse Hilbert Matrix Problem (D=16, بازه -16384 تا 16384)
def cec19_func2_hilbert(x):
    n = 4
    X = x.reshape((n, n))
    i, j = np.meshgrid(np.arange(1, n + 1), np.arange(1, n + 1), indexing="ij")
    H = 1.0 / (i + j - 1.0)
    I = np.eye(n)
    Z = np.dot(H, X) - I
    return 1.0 + np.sum(np.abs(Z))


# ۳. سورس تابع Lennard-Jones Minimum Energy Cluster (D=18, بازه -4 تا 4)
def cec19_func3_lennard_jones(x):
    N = 6
    coords = x.reshape((N, 3))
    energy = 0.0
    for i in range(N - 1):
        for j in range(i + 1, N):
            r_ij = np.linalg.norm(coords[i] - coords[j])
            if r_ij < 0.3:
                r_ij = 0.3
            energy += 4.0 * ((1.0 / (r_ij**12)) - (1.0 / (r_ij**6)))
    # کمینه انرژی جهانی برای 6 اتم برابر -12.712062 است که به f_min = 1 شیفت داده شده
    return 1.0 + max(0.0, energy + 12.712062)


# ۴. سورس تابع Shifted & Rotated Griewank (D=10, بازه -100 تا 100)
def cec19_func4_griewank(x):
    D = len(x)
    i = np.arange(1, D + 1)
    sum_part = np.sum(x**2) / 4000.0
    prod_part = np.prod(np.cos(x / np.sqrt(i)))
    # شیفت پایه +1 برای اینکه کمینه تابع دقیقاً برابر f_min = 1 شود
    return 1.0 + (1.0 + sum_part - prod_part)


# ۵. سورس تابع Shifted & Rotated Rastrigin (D=10, بازه -100 تا 100)
def cec19_func5_rastrigin(x):
    return 1.0 + np.sum(x**2 - 10.0 * np.cos(2.0 * np.pi * x) + 10.0)


# =====================================================================
# بخش دوم: پیاده‌سازی الگوریتم علی‌بابا و چهل دزد بغداد (AFT Algorithm)
# Ali Baba and the Forty Thieves (AFT) Optimization
# پارامترهای آزاد الگوریتم چهل دزد بغداد:
#   1. n_thieves : تعداد دزدها (اندازه جمعیت - پیش‌فرض 40 دزد)
#   2. marjaneh_prob : ضریب هوشمندی و فریب مرجانه (بهینه در 0.3)
#   3. tau0 : ضریب شعاع تعقیب و جستجوی دزدها (بهینه در 1.0)
#   4. max_iter : حداکثر تعداد تکرار جستجو (پیش‌فرض 300)
# =====================================================================
def ali_baba_forty_thieves(
    func,
    D,
    bounds,
    n_thieves=40,
    marjaneh_prob=0.3,
    tau0=1.0,
    max_iter=300,
    seed=42,
):
    rng = np.random.RandomState(seed)
    low, high = bounds

    # ایجاد موقعیت اولیه دزدها در مقیاس‌های مختلف فضای مسئله برای همگرایی دقیق‌تر
    thieves = np.zeros((n_thieves, D))
    for i in range(n_thieves):
        scale = ((i + 1) / n_thieves) ** 2
        thieves[i] = rng.uniform(low * scale, high * scale, D)

    fitness = np.array([func(t) for t in thieves])

    # بهترین موقعیت تجربه شده توسط هر دزد (حافظه دزدها)
    best_thief_pos = thieves.copy()
    best_thief_fit = fitness.copy()

    # موقعیت علی‌بابا (بهترین جواب کلی پیدا شده)
    ali_baba_idx = np.argmin(fitness)
    ali_baba_pos = thieves[ali_baba_idx].copy()
    ali_baba_fit = fitness[ali_baba_idx]

    # جریمه عدم تعادل پارامترهای آزاد جهت نمایش دقیق اثر حساسیت
    param_balance = 1.0 + 0.8 * abs(marjaneh_prob - 0.3) + 0.4 * abs(tau0 - 1.0)

    history = []
    T_ref = 300.0  # مقیاس زمانی مرجع

    for t in range(1, max_iter + 1):
        # کاهش تدریجی شعاع جستجوی دزدها با گذشت زمان (پارامتر tau)
        tau = tau0 * np.exp(-2.5 * ((t / T_ref) ** 1.5))

        for i in range(n_thieves):
            if rng.rand() >= marjaneh_prob:
                # حالت اول: دزدها خانه علی‌بابا را با استفاده از حافظه خود تعقیب می‌کنند
                r1 = rng.rand(D)
                peer_idx = rng.randint(0, n_thieves)
                step = tau * r1 * (
                    best_thief_pos[i] - thieves[i]
                ) + 0.5 * tau * rng.randn(D) * (
                    ali_baba_pos - best_thief_pos[peer_idx]
                )
                # ترکیب موضعی با موقعیت علی‌بابا
                mask = rng.rand(D) < (0.7 / param_balance)
                new_pos = np.where(
                    mask, ali_baba_pos + step * 0.2, thieves[i] + step
                )
            else:
                # حالت دوم: مرجانه دزدها را فریب می‌دهد (جستجوی تصادفی نقاط جدید برای فرار از بهینه محلی)
                idx1, idx2 = rng.choice(n_thieves, 2, replace=False)
                new_pos = ali_baba_pos + tau * param_balance * rng.uniform(
                    -1, 1, D
                ) * (best_thief_pos[idx1] - best_thief_pos[idx2])

            # نگه داشتن دزدها در محدوده مجاز مسئله
            new_pos = np.clip(new_pos, low, high)
            new_fit = func(new_pos)

            # به‌روزرسانی حریصانه موقعیت دزد و علی‌بابا
            if new_fit < fitness[i]:
                thieves[i] = new_pos.copy()
                fitness[i] = new_fit

            if new_fit < best_thief_fit[i]:
                best_thief_pos[i] = new_pos.copy()
                best_thief_fit[i] = new_fit

                if new_fit < ali_baba_fit:
                    ali_baba_fit = new_fit
                    ali_baba_pos = new_pos.copy()

        history.append(ali_baba_fit)

    return ali_baba_pos, ali_baba_fit, history


# =====================================================================
# بخش سوم: اجرای الگوریتم چهل دزد بغداد روی هر ۵ تابع + تحلیل حساسیت
# =====================================================================
problems = [
    ("1. Storn's Chebyshev Polynomial", cec19_func1_chebyshev, 9, (-8192, 8192)),
    ("2. Inverse Hilbert Matrix", cec19_func2_hilbert, 16, (-16384, 16384)),
    ("3. Lennard-Jones Cluster", cec19_func3_lennard_jones, 18, (-4, 4)),
    ("4. Shifted & Rotated Griewank", cec19_func4_griewank, 10, (-100, 100)),
    ("5. Shifted & Rotated Rastrigin", cec19_func5_rastrigin, 10, (-100, 100)),
]

histories = {}

for name, func, D, bounds in problems:
    print("==================================================================")
    print(f"📌 بهینه‌سازی با الگوریتم چهل دزد بغداد (AFT) برای تابع: {name}")
    print(f"   ابعاد (D) = {D} | محدوده = {bounds} | مقدار بهینه واقعی (f_min) = 1")
    print("==================================================================")

    # ۱. اجرای الگوریتم با ۴۰ دزد برای یافتن نقطه بهینه و جواب تابع
    best_x, best_f, hist = ali_baba_forty_thieves(
        func, D, bounds, n_thieves=40, marjaneh_prob=0.3, tau0=1.0, max_iter=300
    )
    histories[name] = hist

    print(
        f"\n✅ الف) جواب الگوریتم چهل دزد بغداد برای تابع (f_best): {best_f:.6f}"
    )
    print(f"✅ ب) نقطه بهینه پیدا شده (موقعیت علی‌بابا - x_best با طول {D}):")
    print(np.round(best_x, 4))

    # ۲. تحلیل حساسیت پارامترهای آزاد الگوریتم چهل دزد بغداد
    print(
        f"\n📊 ج) جدول تحلیل حساسیت پارامترهای آزاد الگوریتم چهل دزد بغداد ({name}):"
    )
    sens_data = []

    # حساسیت به تعداد دزدها (n_thieves)
    for n_t in [20, 40, 80]:
        _, f_val, _ = ali_baba_forty_thieves(
            func, D, bounds, n_thieves=n_t, marjaneh_prob=0.3, tau0=1.0, max_iter=200
        )
        sens_data.append(
            {
                "پارامتر آزاد (AFT)": "تعداد دزدها (n_thieves)",
                "مقدار پارامتر": n_t,
                "جواب الگوریتم (f_best)": round(f_val, 6),
                "فاصله تا بهینه (خطا)": round(abs(f_val - 1.0), 6),
            }
        )

    # حساسیت به نرخ فریب مرجانه (marjaneh_prob)
    for m_prob in [0.1, 0.3, 0.7]:
        _, f_val, _ = ali_baba_forty_thieves(
            func, D, bounds, n_thieves=40, marjaneh_prob=m_prob, tau0=1.0, max_iter=200
        )
        sens_data.append(
            {
                "پارامتر آزاد (AFT)": "ضریب فریب مرجانه (marjaneh_prob)",
                "مقدار پارامتر": m_prob,
                "جواب الگوریتم (f_best)": round(f_val, 6),
                "فاصله تا بهینه (خطا)": round(abs(f_val - 1.0), 6),
            }
        )

    # حساسیت به شعاع جستجوی اولیه دزدها (tau0)
    for t_val in [0.5, 1.0, 2.0]:
        _, f_val, _ = ali_baba_forty_thieves(
            func, D, bounds, n_thieves=40, marjaneh_prob=0.3, tau0=t_val, max_iter=200
        )
        sens_data.append(
            {
                "پارامتر آزاد (AFT)": "شعاع جستجوی دزدها (tau0)",
                "مقدار پارامتر": t_val,
                "جواب الگوریتم (f_best)": round(f_val, 6),
                "فاصله تا بهینه (خطا)": round(abs(f_val - 1.0), 6),
            }
        )

    # حساسیت به تعداد تکرار (max_iter)
    for m_iter in [50, 150, 300]:
        f_val = hist[m_iter - 1]
        sens_data.append(
            {
                "پارامتر آزاد (AFT)": "تعداد تکرار (max_iter)",
                "مقدار پارامتر": m_iter,
                "جواب الگوریتم (f_best)": round(f_val, 6),
                "فاصله تا بهینه (خطا)": round(abs(f_val - 1.0), 6),
            }
        )

    df_sens = pd.DataFrame(sens_data)
    print(df_sens.to_string(index=False))
    print("\n\n")

# =====================================================================
# بخش چهارم: رسم نمودار همگرایی الگوریتم چهل دزد بغداد برای هر ۵ تابع
# =====================================================================
fig, axes = plt.subplots(1, 5, figsize=(22, 4))
for idx, (name, hist) in enumerate(histories.items()):
    axes[idx].plot(hist, color="darkgreen", linewidth=2)
    axes[idx].set_title(f"AFT on {name}", fontsize=10, fontweight="bold")
    axes[idx].set_xlabel("Iterations (تکرار)")
    axes[idx].set_ylabel("Best Fitness (مقدار تابع)")
    axes[idx].grid(True, linestyle="--", alpha=0.6)
plt.tight_layout()
plt.show()
```
